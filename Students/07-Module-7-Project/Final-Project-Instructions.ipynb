**Why this project exists**

A lot of real projects fail because nobody agreed what “done” means, or the demo never becomes something you can run again. We want the opposite.

Picture a small team pitching an ops lead or an investor. You have a **notebook that runs** — it observes, decides, acts, and learns from feedback on a process the firm already cares about. Show us it can save money or cut risk with a **baseline scorecard**, not just a flashy model name. And when you talk about the numbers, **say it in your own words**.

### What counts as an agent

| This works | This does not |
| --- | --- |
| An **LLM**, **RL**, or **hybrid** loop | A chatbot that never acts or gets feedback |
| A fitted sklearn / XGBoost / DAT 5565 Keras model as a **tool** inside the loop | A prediction-only DAT 5565 notebook as the whole project |
| Simulated data, public data, or a free API ([public-apis](https://github.com/public-apis/public-apis), prefer No Auth) | Confidential employer files |

Tell us **who releases the frequent action** (a system or a person). If you go hybrid, say who does what.


### What we expect

1. **A real business process** — something the firm already runs. “Solved” is a **number they already track** (cost, SLA, fill rate, false holds, …).
2. **A family that fits the work** — LLM, RL, or hybrid for a reason tied to the job, not the brand.
3. **An artifact you can rerun** — a notebook (or script) for a new day, ticket, or feed.
4. **A fair baseline and a scorecard** — task success, cost, latency, safety/grounding, business KPI.
5. **Results explained in your own words** — what the table means, what surprised you, whether you would deploy. This is a big part of the grade.
6. **Governance** — when a person steps in, what you log, what must stay out of a public model, buy vs build.
7. **A live talk** — about **7 minutes** in Session 13 (not a 15-minute Zoom recording). Anyone on the team should be able to walk through the scorecard and say who releases.
8. **Named dataset** — synthetic/toy earns less credit; live API / public earns more.
9. **Approaches you combine** — fewer distinct approaches earn less credit (see the rubric ladder).

Use whatever stack fits (Python notebooks are fine). TensorFlow-Keras is **not** required. PyTorch is fine when it helps.

### How to work through it

- **Pick a process** — refunds/claims, offers, restock, document + number, handoff, fraud hold/release, FX sleeve, pricing with a policy check — or invent your own.
- **Write a short proposal** — half a page to one page. We mainly use it for feedback unless Canvas scores it. See `Project-Proposal.md`.
- **Build the loop** — Observe → Decide → Act → Feedback for *your* process (please do not paste the Session 1 definition).
- **Evaluate** — run the agent and a fair baseline; fill the scorecard; **explain the results in your own words**.
- **Write the report** — fill `Report-Template.md` (Markdown, Word, or PDF). One member submits for the group.
- **Present** — ~7 minutes: loop → dataset/approaches → **what the scorecard means** → governance (and who releases).
