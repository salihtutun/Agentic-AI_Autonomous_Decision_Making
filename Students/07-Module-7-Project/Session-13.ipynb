```python
# Fill this in before you present. Simulated, public, a free public API, or an ML model used as a tool. Never confidential.
PROJECT = {
    "problem": "",  # one real business process, not a model name
    "data": "",  # named dataset or API + tier (synthetic / public / live API)
    "approaches": "",  # distinct approaches combined, e.g. "baseline + Q + veto"
    "family": "",  # LLM, RL, or hybrid
    "who_releases": "",  # system or person that releases the frequent action
    "loop": {
        "observe": "",
        "decide": "",
        "act": "",
        "feedback": "",
    },
    "baseline": "",  # the simple rule or chatbot you beat, or failed to beat
    "metrics": {
        "task_success": None,
        "cost": None,
        "latency_s": None,
        "safety_or_grounding": None,
        "business_kpi": None,
    },
    "governance": {
        "human_checkpoint": "",
        "audit": "",
        "privacy": "",
        "buy_vs_build": "",
    },
}


def ready(project):
    # A prediction-only model is not ready: dataset, approaches, loop, baseline, and KPI
    # have to be filled before the group presents.
    missing = []
    if not project["problem"]:
        missing.append("business problem")
    if not project["data"]:
        missing.append("named dataset/API (+ tier)")
    if not project["approaches"]:
        missing.append("approaches combined")
    if project["family"] not in {"LLM", "RL", "hybrid"}:
        missing.append("family")
    if not project["who_releases"]:
        missing.append("who releases")
    if any(not text for text in project["loop"].values()):
        missing.append("agent loop")
    if not project["baseline"]:
        missing.append("baseline")
    if any(value is None for value in project["metrics"].values()):
        missing.append("metrics")
    if any(not text for text in project["governance"].values()):
        missing.append("governance")
    if missing:
        print("Still missing:", ", ".join(missing))
        return False
    print("Checklist complete. Practice the 7 minutes: loop → dataset/approaches → scorecard → governance.")
    return True


ready(PROJECT)
```
