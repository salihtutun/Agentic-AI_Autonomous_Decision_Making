
````python
# ============================================================
# PART 4 — FILL the memo, then run review_memo
# ============================================================

MEMO = {
    "process": "Dynamic price with a promo email",
    "mdp_design": {
        "states": "",  # <<< FILL (from Part 1)
        "actions": [],  # <<< FILL ["low", "high"]
        "reward": "",  # <<< FILL (margin / profit / revenue)
    },
    "llm_design": {
        "goal": "",  # <<< FILL (from Part 3)
        "human_handoff": "",  # <<< FILL
    },
    "recommendation": "",  # <<< FILL: LLM, RL, hybrid, or do not automate
    "because": "",  # <<< FILL: ≥ 40 characters
}

def review_memo(memo):
    """Student-facing checker. Prints PASS or a short fix list."""
    issues = []
    mdp = memo.get("mdp_design", {})
    llm = memo.get("llm_design", {})
    states = (mdp.get("states") or "").strip()
    actions = mdp.get("actions") or []
    reward = (mdp.get("reward") or "").lower()
    goal = (llm.get("goal") or "").strip()
    handoff = (llm.get("human_handoff") or "").strip()
    rec = (memo.get("recommendation") or "").strip()
    because = (memo.get("because") or "").strip()

    if len(states) < 10:
        issues.append("MDP states: say what you observe before choosing today's price (last demand).")
    if set(a.lower() for a in actions) < {"low", "high"}:
        issues.append('MDP actions: include both "low" and "high".')
    if "margin" not in reward and "profit" not in reward and "revenue" not in reward:
        issues.append("MDP reward: mention margin, profit, or revenue.")
    if len(goal) < 8:
        issues.append("LLM design needs a goal for the promo email.")
    if len(handoff) < 8:
        issues.append("LLM design needs a human_handoff (who accepts the override).")
    if rec not in {"LLM", "RL", "hybrid", "do not automate"}:
        issues.append("recommendation must be LLM, RL, hybrid, or do not automate.")
    if rec == "LLM":
        issues.append("Last period's demand is numeric. Do not give the daily price itself to an LLM agent.")
    if len(because) < 40:
        issues.append("because is too short (need ≥ 40 characters a manager can read aloud).")

    if issues:
        print("NOT YET")
        for issue in issues:
            print("-", issue)
        return False

    print("PASS")
    print(rec + ":", because)
    return True


review_memo(MEMO)
````
