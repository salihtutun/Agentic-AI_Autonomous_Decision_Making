Individual. Due after Week 3 Wednesday.

New process, not replenishment. Each visit the site shows price **low** or **high**. The state you can measure is last period's demand, **low** or **high**. A promo email sometimes arrives and is not in that state.

Fill `MEMO`. `review_memo` must print PASS.


```python
MEMO = {
    "process": "Dynamic price with a promo email",
    "mdp_design": {
        "states": "",
        "actions": [],
        "reward": "",
    },
    "llm_design": {
        "goal": "",
        "human_handoff": "",
    },
    "recommendation": "",  # LLM, RL, hybrid, or do not automate
    "because": "",
}


def review_memo(memo):
    issues = []
    reward = memo["mdp_design"]["reward"].lower()
    if not memo["mdp_design"]["states"] or len(memo["mdp_design"]["actions"]) < 2:
        issues.append("MDP needs a state and both prices as actions.")
    if "margin" not in reward and "profit" not in reward and "revenue" not in reward:
        issues.append("Reward must mention margin, profit, or revenue.")
    if memo["recommendation"] not in {"LLM", "RL", "hybrid", "do not automate"}:
        issues.append("Recommendation is missing.")
    if memo["recommendation"] == "LLM":
        issues.append("Last period's demand is numeric. Do not give the price itself to an LLM agent.")
    if not memo["llm_design"]["human_handoff"]:
        issues.append("The promo email needs a human handoff.")
    if len(memo["because"]) < 40:
        issues.append("because is too short.")
    if issues:
        print("NOT YET")
        for issue in issues:
            print("-", issue)
        return False
    print("PASS")
    return True


review_memo(MEMO)
```
