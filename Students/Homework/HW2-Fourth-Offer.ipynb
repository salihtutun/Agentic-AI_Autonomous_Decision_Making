Session 4 practiced ε-greedy on real Frankfurter FX sleeves. This homework is the same rule on a clean fourth promotional offer, so you are not copying class numbers.

Session 4 used offers A, B, and C. Add offer **D** with true accept rate **0.50**. Keep epsilon at 0.10 and the seed at 7. Report which offer the values pick, and one sentence on whether D changed the decision.


```python
import numpy as np

OFFERS = ["A", "B", "C"]  # TODO: include "D"
TRUE_RATE = {"A": 0.20, "B": 0.55, "C": 0.35}  # TODO: D is 0.50
RNG = np.random.default_rng(7)
NOTE = ""  # TODO: at least 40 characters. Did the learned best offer change from B?


def customer_accepts(offer):
    return int(RNG.random() < TRUE_RATE[offer])


def epsilon_greedy(values, epsilon, rng):
    if rng.random() < epsilon:
        return int(rng.integers(0, len(values)))
    return int(np.argmax(values))


def run_bandit(epsilon=0.10, n_visits=800):
    values = np.zeros(len(OFFERS))
    counts = np.zeros(len(OFFERS))
    for _ in range(n_visits):
        index = epsilon_greedy(values, epsilon, RNG)
        reward = customer_accepts(OFFERS[index])
        counts[index] += 1
        values[index] += (reward - values[index]) / counts[index]
    return values, counts


def check():
    if "D" not in OFFERS or "D" not in TRUE_RATE:
        print("NOT YET add offer D")
        return
    if len(NOTE) < 40:
        print("NOT YET write NOTE")
        return
    values, counts = run_bandit()
    learned = OFFERS[int(np.argmax(values))]
    print("learned", learned, "counts", counts.astype(int).tolist())
    print(NOTE)
    print("B is still the true best rate. Say whether your run agreed.")


check()
```
