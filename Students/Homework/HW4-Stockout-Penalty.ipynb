```python
import numpy as np

CAPACITY = 4
GAMMA = 0.95
HOLD, MARGIN = 0.2, 2.0
NOTE = ""  # TODO: at least 80 characters. What happened to orders when the stockout penalty became 0?


def run(stockout_penalty, seed=4):
    rng = np.random.default_rng(seed)
    q_table = np.zeros((CAPACITY + 1, 3))

    def step(stock, order):
        demand = int(rng.integers(0, 3))
        started = min(CAPACITY, stock + order)
        sales = min(started, demand)
        ending = started - sales
        missed = max(0, demand - started)
        pay = MARGIN * sales - HOLD * ending - stockout_penalty * missed
        return ending, pay

    for _ in range(400):
        stock = int(rng.integers(0, CAPACITY + 1))
        for _day in range(15):
            action = int(rng.integers(0, 3)) if rng.random() < 0.15 else int(np.argmax(q_table[stock]))
            ending, pay = step(stock, action)
            target = pay + GAMMA * np.max(q_table[ending])
            q_table[stock, action] += 0.1 * (target - q_table[stock, action])
            stock = ending
    return np.argmax(q_table, axis=1)


with_penalty = run(1.5)
without_penalty = run(0.0)
print("order by stock, penalty on ", with_penalty.tolist())
print("order by stock, penalty off", without_penalty.tolist())
print("PASS" if len(NOTE) >= 80 else "NOT YET write at least 80 characters in NOTE")
```
