## How to read one scorecard

Use the same rows for both families. A higher task-success number is not a reason to deploy.

- **Task success** — share of cases finished without a person rewriting the action.
- **Cost and latency** — what one case costs and how long the decide step takes. The RL pilot is cheaper and faster because it does not read an email.
- **Safety** — share of actions inside the rule you would defend (policy for the language agent, capacity and the veto for the order agent).
- **Grounding** — share of language replies that cite a real chunk. It applies to the LLM pilot. `None` on the RL pilot means the job is not a document, not that the RL pilot failed.
- **KPI against a baseline** — stockout rate versus base-stock (0.14 here). Beating the baseline is the business result. Losing to it is an acceptable finding if you say so.

**How to recommend.** If grounding or safety is too weak to auto-release that family, that family does not release the action. Here, grounding 0.62 means about four replies in ten have no real citation, so the language agent must not auto-refund or auto-override. The order agent can still own the repeated quantity if it beats base-stock and stays inside the veto. That split is a **hybrid**: the fast policy releases the daily order, and a person releases any email override. Choose **do not automate** when the action that would go live fails safety, or when neither family beats the baseline and you cannot name a checkpoint.
